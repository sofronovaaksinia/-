## Задание 2

```python
def div_mod(dividend, divisor):
  """Возвращает частное и остаток от деления dividend на divisor."""
  if divisor == 0:
    raise ValueError("Делитель не может быть равен нулю.")
  quotient = dividend // divisor
  remainder = dividend % divisor
  return quotient, remainder
q, r = div_mod(17, 5)
print(f"Частное: {q}, Остаток: {r}")
raw_result = div_mod(17, 5)
print("Тип возвращаемого объекта:", type(raw_result))
```

В Python функция всегда возвращает ровно один объект. Когда через запятую указано несколько значений, интерпретатор автоматически упаковывает их в единый объект типа кортеж. При присваивании вида q, r = ... срабатывает обратный механизм — распаковка кортежа в отдельные переменные.

## Задание 3

```python
def hanoi(n, source="A", target="C", auxiliary="B", print_moves=True):
  """Рекурсивно решает задачу о Ханойских башнях и возвращает количество ходов.
    n — количество дисков
    source — исходный стержень
    target — целевой стержень
    auxiliary — вспомогательный стержень
    """
  if n <= 0:
    return 0
  if n == 1:
    if print_moves:
      print(f"Переместить диск 1 со стержня {source} на стержень {target}")
    return 1
  moves_before = hanoi(n - 1, source, auxiliary, target, print_moves)
  if print_moves:
    print(f"Переместить диск {n} со стержня {source} на стержень {target}")
  moves_after = hanoi(n - 1, auxiliary, target, source, print_moves)
  return moves_before + 1 + moves_after
print("--- Последовательность ходов для 3 дисков ---")
m3 = hanoi(3, print_moves=True)
m4 = hanoi(4, print_moves=False)
m5 = hanoi(5, print_moves=False)
print(f"\nЧисло ходов для 3 дисков: {m3}")
print(f"Число ходов для 4 дисков: {m4}")
print(f"Число ходов для 5 дисков: {m5}")
```

Вывод о росте сложности:
• Для n = 3 требуется 7 ходо (2^3 - 1).
• Для n = 4 требуется 15 ходов (2^4 - 1).
• Для n = 5 требуется 31 ход (2^5 - 1).
Временная сложность алгоритма является экспоненциальной — O(2^n). При добавлении каждого нового диска количество операций удваивается (плюс 1 ход), поэтому при больших n задача становится вычислительно невыполнимой за разумное время.

  return counter
