## Задание

Напишите функцию, которая печатает натуральные числа от **1** до **N** с шагом **step**. Числа печатаются через пробел в строку. **N** и **step** задаются как параметры функции.


## Решение

### Объявление функции

```python
def print_numbers(n, step):
  for i in range(1, n+1, step):
    print(i, end='')
  print()
```

### Демонстрация использования функции

```python
N, step = 10, 2
print(f'Натуральные числа от 1 до {N} с шагом {step}:')
print_numbers(N, step)
N, step = 25, 5
print(f'Натуральные числа от 1 до {N} с шагом {step}:')
print_numbers(N, step)
N, step = 14, 3
print(f'Натуральные числа от 1 до {N} с шагом {step}:')
print_numbers(N, step)
```
