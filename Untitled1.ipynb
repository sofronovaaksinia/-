## Ответы на контрольные вопросы

1. Переменная — это именованная ссылка на объект в оперативной памяти компьютера. При присваивании (=) в памяти создаётся объект со значением, а имя переменной связывается с адресом этого объекта.
Корректные имена: user_age, totalSum, item2.
Некорректное имя: 2item (имя переменной не может начинаться с цифры).
2. • Целое число (int): x = 10 — <class 'int'>.
• Вещественное число (float): y = 3.14 — <class 'float'>.
• Строка (str): s = "текст" — <class 'str'>.
• Логический тип (bool): flag = True — <class 'bool'>.
Динамическая типизация означает, что тип привязан к самому значению, а не к имени переменной. Переменная может в любой момент начать ссылаться на объект другого типа.
3. • Одна переменная: print(x) — выводит значение и переносит строку.
• Через запятую: print("Возраст:", 20) — автоматически ставит пробел между аргументами и сам преобразует числа в текст.
• Форматированный вывод (f-строка): print(f"Возраст: {age}") — подставляет значения внутрь шаблона. В отличие от конкатенации ("Возраст: " + str(age)), f-строка не требует ручного перевода чисел в строки и работает быстрее.
4. Функция возвращает класс переданного объекта.
a = 5
b = "5"
c = False
print(type(a), type(b), type(c))
Вывод: <class 'int'> <class 'str'> <class 'bool'>
Запись <class 'int'> означает, что объект является экземпляром класса целых чисел.
5. • "+", "-", "*", "**": если оба числа int, результат int (кроме возведения в отрицательную степень). Если хотя бы одно float — результат float.
• "/": всегда возвращает float.
• "//" и "%": возвращают int для целых чисел и float, если один из операндов дробный.
6. Оператор "/" выполняет точное деление, а "//" округляет результат в меньшую сторону.
Положительные числа: 7 / 2 = 3.5, но 7 // 2 = 3.
Отрицательные числа: -7 / 2 = -3.5, а -7 // 2 = -4.
7. Для явного приведения используются функции int(), float(), str().
Примеры: int("10") = 10, float(5) = 5.0, str(100) = "100".
Ошибка ValueError возникнет при попытке перевести в число строку, не содержащую корректного числа, например int("привет") или int("3.14") (строку с точкой нужно сначала переводить через float()).
8. Любая комбинация значений и операторов вычисляется интерпретатором в итоговый объект, который можно сразу сохранить или вывести.
• Арифметическое: res = 2 + 3 * 4 (вернёт 14).
• Сравнение: is_bigger = 10 > 5 (вернёт True).
• Логическое: print(True and False) (выведет False).
9. • Сравнение (==, !=, <, >, <=, >=) проверяет соотношение значений и возвращает bool.
• Логические операторы (not, and, or) комбинируют логические значения.
Пример: 5 > 3 and not (2 == 4 or 1 > 5). Сначала вычисляются скобки (False or False — False), затем not False — True, затем 5 > 3 (True), и в конце True and True — True.
10. Неявное приведение срабатывает для совместимых типов, например int и float (5 + 2.5 автоматически преобразует 5 в 5.0 и вернёт 7.5). Также bool неявно приводится к int (True + 2 = 3)
Для несовместимых типов (str и int) сложение вызовет ошибку TypeError, поэтому требуется явное приведение ("Число: " + str(5)). Исключение — умножение строки на целое число ("а" * 3 = "ааа").
11. • Через запятую: print("X =", x) — удобно для быстрой отладки.
• f-строки: print(f"X = {x:.2f}") — самый современный и удобный способ для сложного форматирования текста.
• Метод .format(): print("X = {}".format(x)) — удобен, если шаблон строки задаётся заранее отдельно от данных.
• Конкатенация: print("X = " + str(x)) — подходит только при склеивании уже готовых строк.
12. Приоритет от высшего к низшему: скобки (), возведение в степень **, унарные плюс/минус, умножение и деление (*, /, //, %), сложение и вычитание (+, -).
Пример: 2 + 3 * 4 даст 14, но (2 + 3) * 4 изменит порядок и даст 20.
13. • Одновременное присваивание: x, y, z = 1, 2, 3.
• Обмен без третьей переменной: a, b = b, a.
• Распаковка списка/кортежа: coords = (10, 20) — x, y = coords.
14. Числа (int, float) и строки (str) являются неизменяемыми. После создания объекта в памяти изменить его содержимое нельзя.
При попытке «изменить» переменную создаётся новый объект по новому адресу в памяти, что подтверждает функция id():
s = "код"
print(id(s))
s = s + "!"
print(id(s))
15. count = 4                  # Создаём целое число (int)

price = 12.5               # Создаём дробное число (float)

item = "Ручка"             # Создаём строку (str)

total = count * price      # Умножаем int на float, результат становится float (50.0)

print(f"{item}: {total}")  # Выводим форматированную строку

print(type(total))         # Проверяем тип результата -> <class 'float'>
