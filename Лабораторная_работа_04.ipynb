## Задание

В файле data_v1.txt содержится последовательность символов, содержащая буквы латинского алфавита и десятичные цифры. Необходимо решить следующую задачу:
Найдите количество чисел кратных 10 (под числом понимается непрерывная последовательность цифр). В ответе укажите количество таких чисел и их среднее арифметическое.

## Анализ задания, получение данных

Файл data_v1.txt расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

```python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/main/data/data_v1.txt
```

Реализуем чтение всего файла в строку с использованием метоа `open()` и `read()`:

```python
f=open('data_v1.txt')
s=f.read().strip()
print(len(s))
```

## Обработка символьной последовательности

```python
# Вывод первых 10 символов строки с одновременным анализом буква это или цифра
for i in range(10):
  c = s[i]
  if c.isdigit():
    t = 'цифра'
  else:
    t='буква'
  print(c, t)
```

## Решение задачи

Будем посимвольно обходимть строку и накапливать символы-цифры в строке `target`. Если цифровая последовательность прервется, то конвертируем `target` в число `int`, проверяем, делится ли число на 10 без остатка, и сохраняем в список для последующего подсчета среднего арифметическог.

```python
target=''
k=0
for c in s:
  if c.isdigit():
    target+=c
  else:
    if target!='':
      if int(target)%10==0:
        k+=1
      target=''
if target!='' and int(target)%10==0:
  k+=1
print(k)
```

Подсчитываем среднее арифметическое найденных чисел.

```python
target=''
numbers=[]
for c in s:
  if c.isdigit():
    target+=c
  else:
    if target!='':
      x=int(target)
      if x%10==0:
        numbers.append(x)
      target=''
if target!='':
  x=int(target)
  if x%10==0:
    numbers.append(x)
average=sum(numbers) / len(numbers) if len(numbers)>0 else 0
print (average)
```
