## Задание 2

```python
n = int(input("Введите количество чисел n: "))
pos_count = 0
neg_count = 0
zero_count = 0
for i in range(n):
    num = float(input(f"Введите число {i + 1}: "))
    if num > 0:
        pos_count += 1
    elif num < 0:
        neg_count += 1
    else:
        zero_count += 1
print(f"Положительных: {pos_count}")
print(f"Отрицательных: {neg_count}")
print(f"Нулей: {zero_count}")
```

## Задание 3

```python
import math
def sum_proper_divisors(num):
    if num <= 1:
        return 0
    total = 1  # 1 всегда является делителем для чисел > 1
    limit = int(math.sqrt(num))
    for i in range(2, limit + 1):
        if num % i == 0:
            total += i
            pair = num // i
            if pair != i:  # исключаем дублирование для полных квадратов
                total += pair
    return total
n = int(input("Введите верхнюю границу n: "))
found = False
for a in range(2, n + 1):
    b = sum_proper_divisors(a)
    # Проверяем b > a, чтобы не выводить одну пару дважды и исключить совершенные числа (a == b)
    if a < b <= n:
        if sum_proper_divisors(b) == a:
            print(f"Дружественная пара: {a} и {b}")
            found = True
if not found:
    print("Дружественных пар в заданном диапазоне не найдено.")
```

## Ответы на вопросы

1. for используется, когда заранее известно количество шагов или нужно перебрать элементы готовой коллекции.
while применяется, когда число итераций заранее неизвестно и зависит от выполнения логического условия.
2. Блок else выполняется только в том случае, если цикл завершился штатно и не был прерван оператором break.

(Пример 1: else срабатывает)

for x in:
    if x == 5:
        break
else:
    print("Цикл завершён без break")
(Пример 2: else НЕ срабатывает)
for x in:
    if x == 5:
        break
else:
    print("Это сообщение не выведется")

3. break полностью и досрочно прерывает выполнение всего цикла.
continue прерывает только текущую итерацию и сразу переходит к проверке условия следующей итерации.
i = 0
while i < 5:
    i += 1
    if i == 2:
        continue  (пропускаем вывод числа 2, идём на i = 3)
    if i == 4:
        break     (при i = 4 полностью выходим из цикла)
    print(i)      # выведет только 1 и 3

4. • Бесконечный цикл:
Ошибка: i = 1; while i < 5: print(i)
Исправление: добавить i += 1 внутри тела цикла.
• Неверный диапазон range:
Ошибка: for i in range(1, 5): print(i) (выведет от 1 до 4 вместо 5).
Исправление: for i in range(1, 6): print(i).
• Неправильный шаг для убывающего диапазона:
Ошибка: for i in range(10, 1): print(i).
Исправление: указать отрицательный шаг for i in range(10, 0, -1): print(i).

6. and возвращает True, если истинны оба выражения; or — если истинно хотя бы одно; not инвертирует логическое значение. Приоритет: not — and — or.
Выражение: (x % 2 == 0 and 10 <= x <= 20) or (x % 2 != 0 and x > 100)
7. Порядок убывания приоритета: скобки () — возведение в степень ** — унарный минус — умножение и деление *, /, //, % — сложение и вычитание +, -.
Примеры:
	• 2 + 3 * 4 ** 2 —> 2 + 3 * 16 —> 50
	• (2 + 3) * 4 —> 20
	• 17 // 5 + 17 % 5 —> 3 + 2 —> 5
8. / — истинное деление, всегда возвращает вещественный тип float.
// — целочисленное деление, возвращает int (если оба операнда int) или float (если хотя бы один float).
print(7 / 2, type(7 / 2))     (3.5 <class 'float'>)
print(7 // 2, type(7 // 2))   (3 <class 'int'>)
print(-7 // 2, type(-7 // 2)) (-4 <class 'int'>)
9. • min([3, 1, 5]) — 1 (принимает итерируемый объект или несколько аргументов; возвращает наименьший элемент того же типа).
• max(3, 1, 5) — 5 (возвращает наибольший элемент).
• sum([1, 2, 3.5]) — 6.5 (принимает коллекцию чисел; возвращает их сумму int или float).
• abs(-4.2) — 4.2 (принимает число int/float/complex; возвращает его модуль).
• round(3.14159, 2) — 3.14 (принимает число и необязательное число знаков; возвращает float или int, если второй аргумент не указан).
10. Синтаксис: range(start, stop, step). Верхняя граница stop никогда не включается в диапазон.
• Шаг 1: range(1, 5) — 1, 2, 3, 4
• Отрицательный шаг: range(5, 0, -1) — 5, 4, 3, 2, 1
• Произвольный шаг: range(0, 10, 3) — 0, 3, 6, 9
11. Подключается инструкцией import math. Без предварительного импорта вызов sqrt(9) приведёт к ошибке NameError: name 'sqrt' is not defined.

Метод Герона: плюсы — позволяет управлять точностью и изучать численные методы; минусы — медленнее и сложнее в написании.
