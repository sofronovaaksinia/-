Теперь ответим на поставленные вопросы, используя строку `mystr`

```python
mystr = open('pr6.txt').read().strip()
# mystr='ASD.DFGUIO.OOOAEEERUY.UYTEE'
glas = 'AEIOUY'

# заменим гласные на символ *
for c in glas:
  mystr = mystr.replace(c, '*')

# порежем строку по символу *
sub_str = mystr.split('.')
max_len = 0

# выполним обход всех подстрок ...
for s in sub_str:
  z = s.split('*')
  # ... и проанализируем каждую подстроку
  for i in range(len(z)-7):
    t = ''.join(z[i:i+8])
    if max_len < len(t):
      max_len = len(t) + len(z[i:i+8])-1
      print(t)
      print(z[i:i+8])

print(f'Максимальная подходящая подстрока: {max_len}')
```

## Задание по варианту 7

Текстовый файл pr6_1.txt состоит не более чем из 106 символов и
содержит только заглавные буквы латинского алфавита и
десятичные цифры. Найдите два числа: количество букв и
количество четных цифр.

```python
with open('pr6_1.txt', 'r', encoding='utf-8') as file:
    text = file.read().strip()
letters_count = 0
even_digits_count = 0
for char in text:
    if char.isalpha():
        letters_count += 1
    elif char.isdigit() and int(char) % 2 == 0:
        even_digits_count += 1
print(f"Количество букв: {letters_count}")
print(f"Количество четных цифр: {even_digits_count}")
```
