## Задание 2

```python
def find_most_expensive_product(products):
    """
    Принимает список кортежей вида (название, цена, количество).
    Возвращает название товара с максимальной общей стоимостью.
    Если список пуст, возвращает None.
    """
    if not products:
        return None
    max_cost = -1
    best_product_name = ""
    for name, price, quantity in products:
        total_cost = price * quantity
        if total_cost > max_cost:
            max_cost = total_cost
            best_product_name = name
    return best_product_name
store_items = [
    ("Ноутбук", 55000, 2),
    ("Смартфон", 30000, 4),
    ("Наушники", 5000, 10),
    ("Монитор", 22000, 3)
]
result = find_most_expensive_product(store_items)
print(f"Товар с максимальной общей стоимостью: {result}")
```

## Задание 3

```python
def group_by_first_letter(words):
    """
    Группирует список строк по их первой букве (без учёта регистра).
    Возвращает словарь, где ключ — заглавная первая буква,
    а значение — список соответствующих слов.
    """
    grouped = {}
    for word in words:
        if not word:
            continue
        key = word[0].upper()
        if key not in grouped:
            grouped[key] = []
        grouped[key].append(word)
    return grouped
cities = ["Ставрополь", "Самара", "Москва", "Мурманск", "Казань", "Сочи", "Краснодар"]
grouped_cities = group_by_first_letter(cities)
for letter, city_list in grouped_cities.items():
    print(f"{letter}: {city_list}")
```

## Ответы на контрольные вопросы

1. Изменяемые объекты можно модифицировать после создания без изменения их идентификатора в памяти. Пример: list.
Неизменяемые объекты нельзя изменить после создания; любая операция «изменения» на самом деле создаёт новый объект в памяти. Примеры: str, tuple, range.
2. В списке хранятся ссылки на объекты, и Python позволяет перезаписывать эти ссылки по индексам или менять размер контейнера. Строка представляет собой единый непрерывный блок символов в памяти, оптимизированный для быстрого чтения и хеширования.
При попытке выполнить s[0] = 'A' возникнет исключение TypeError: 'str' object does not support item assignment.
3. Доступ осуществляется через квадратные скобки: lst[i], где i — целое число.
При обращении к индексу за пределами диапазона [-len(lst), len(lst) - 1] возникнет исключение IndexError: list index out of range.
4. Отрицательная индексация позволяет обращаться к элементам с конца списка, не вычисляя его длину.
Индекс -1 соответствует последнему элементу списка, индекс -2 — предпоследнему.
5. Срез a[start:stop:step] возвращает новый подсписок, начиная с индекса start до stop с шагом step.
• Верхняя граница не включается, потому что это позволяет удобно вычислять длину среза (stop - start) и стыковать соседние срезы без дублирования элементов: a[:4] + a[4:] == a.
6. Изменение элемента: a[2] = 100 (заменяет строго один элемент на один новый объект).
• Замена через срез: a[1:3] = [10, 20, 30] принимает любую итерируемую последовательность и может изменить общую длину списка.
7. • append(x) — добавляет один объект x в конец списка. Уместен при стандартном накоплении элементов.
• extend(iterable) — распаковывает переданную последовательность и добавляет все её элементы по отдельности в конец списка. Уместен для объединения списков на месте.
• insert(i, x) — вставляет элемент x перед индексом i, сдвигая остальные элементы вправо. Уместен, когда важна конкретная позиция вставки.
8. • lst.remove(x) — удаляет первое вхождение по значению x. Возвращает None.
• lst.pop(i) — удаляет элемент по индексу i и возвращает удалённый элемент.
• del lst[i] (или del lst[1:3]) — инструкция языка, удаляющая элемент или целый срез по индексу. Ничего не возвращает.
9. Произойдёт ошибка ValueError: list.remove(x): x not in list.
Сделать удаление безопасным можно двумя способами:
	1. Предварительная проверка: if x in lst: lst.remove(x)
	2. Блок перехвата исключения: try: lst.remove(x) except ValueError: pass
10. Проверка наличия: оператор x in lst (возвращает True или False).
Поиск индекса первого вхождения: метод lst.index(x). Если элемента нет, он выбрасывает ValueError.
Корректная обработка: проверять условие if x in lst: перед вызовом .index(x) либо оборачивать вызов в блок try ... except ValueError.
11. • lst.sort() — метод списка, который сортирует его на месте и возвращает None. Работает только со списками.
• sorted(iterable) — встроенная функция, которая не трогает исходный объект, а возвращает новый отсортированный список. Принимает любую итерируемую коллекцию.
12. Переменные в Python хранят ссылки на объекты. Запись b = a просто копирует ссылку, поэтому обе переменные указывают на один и тот же список в памяти.
Способы создания поверхностной копии: метод b = a.copy(), срез b = a[:] или конструктор b = list(a).
13. • Конкатенация (a + b) — создаёт новый список, соединяя элементы a и b.
• Повторение (a * n) — создаёт новый список, дублируя содержимое a указанное число раз n.
• Сравнение (==, !=, <, >) — выполняется лексикографически.
• Проверка вхождения (x in a, x not in a) — последовательно проверяет, равен ли хотя бы один элемент списка значению x, возвращая bool.
14. В Python аргументы передаются по ссылке на объект. Если передать список в функцию и изменить его внутри, это вызовет побочный эффект — данные изменятся и в вызывающем коде, что затрудняет отладку.
Создание копии или нового списка внутри функции с последующим возвратом через return делает функцию предсказуемой, «чистой» и безопасной для повторного использования.
